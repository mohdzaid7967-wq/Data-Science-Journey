```python
from secrete_key import GROQ_API_KEY
import os

os.environ["GROQ_API_KEY"] = GROQ_API_KEY

from langchain_groq import ChatGroq

llm = ChatGroq(
    model="openai/gpt-oss-120b",
    temperature=0.6
)

response = llm.invoke(
    "I want to open a restaurant for Indian food. Suggest a fancy name for this."
)

print(response.content)
```

```python
from langchain_core.output_parsers import StrOutputParser


parser = StrOutputParser()


# Chain 1: Generate restaurant name
name_prompt = PromptTemplate(
    input_variables=["cuisine"],
    template="I want to open a restaurant for {cuisine} food. Suggest a fancy restaurant name."
)

name_chain = name_prompt | llm | parser


# Chain 2: Generate food items
food_prompt = PromptTemplate(
    input_variables=["restaurant_name"],
    template="Suggest 5 popular food items for a restaurant named {restaurant_name}."
)

food_items_chain = food_prompt | llm | parser


# Sequential chain
name = name_chain.invoke({"cuisine": "India"})

food_items = food_items_chain.invoke({
    "restaurant_name": name
})


print("Restaurant Name:")
print(name)

print("\nFood Items:")
print(food_items)
```

```python
from langchain.agents import create_agent
from langchain_community.tools import WikipediaQueryRun
from langchain_community.utilities import WikipediaAPIWrapper

# Wikipedia tool
wikipedia = WikipediaQueryRun(
    api_wrapper=WikipediaAPIWrapper()
)

# Create agent
agent = create_agent(
    model=llm,
    tools=[wikipedia],
    system_prompt="You are a helpful assistant. Use Wikipedia when you need factual information."
)

# Run agent
response = agent.invoke({
    "messages": [
        {
            "role": "user",
            "content": "When was Elon Musk born? What is his age right now in 2028?"
        }
    ]
})

print(response["messages"][-1].content)
```
