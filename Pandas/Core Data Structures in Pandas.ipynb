```python
import pandas as pd
```

```python
# Series — 1D Labeled Array

s = pd.Series([10, 20, 30, 40])
print(s)
```

```python
# give custom index
s = pd.Series([10, 20, 30], index=["a", "b", "c"])
print(s)
```

```python
# DataFrame — 2D Labeled Table
data = {
    "name": ["Alice", "Bob", "Charlie"],
    "age": [25, 30, 35],
    "city": ["Delhi", "Mumbai", "Bangalore"]
}

df = pd.DataFrame(data)
print(df)
```

```python
df.index = ["a", "b", "c"]
df.columns = ["Name", "Age", "City"]
```
