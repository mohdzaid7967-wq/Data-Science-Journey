```python
import pandas as pd
import numpy as np
import tensorflow as tf
from tensorflow import keras
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv("insurance_data.csv")
df.head()
```

```python
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(df[['age','affordibility']], df[['bought_insurance']], test_size=0.2,random_state=2)
```

```python
X_train_scalled = X_train.copy()
X_train_scalled['age'] = X_train_scalled['age'] / 100

X_test_scalled = X_test.copy()
X_test_scalled['affordibility'] = X_test_scalled['affordibility'] / 100
```

```python
X_train_scalled
```

```python
model = keras.Sequential([
    keras.layers.Dense
    (
    1, 
    input_shape=(2,),
    activation = 'sigmoid',
    kernel_initializer='ones',
    bias_initializer='zeros'
    )
])

model.compile(
    optimizer='adam',
    loss='binary_crossentropy',
    metrics=['accuracy']
)

model.fit(X_train_scalled, y_train, epochs=5000)
```

```python
model.evaluate(X_test_scalled, y_test)
```

```python
def sigmoid(x):
    return 1 / (1 + np.exp(-x))

def prediction_function(age, affordability):
    weighted_sum = coef[0][0] * age + coef[0][1] * affordability + intercept[0]
    return sigmoid(weighted_sum)
```

```python
def log_loss(y_true, y_predicted):
    epsilon = 1e-15
    y_predicted_new = [max(i,epsilon) for i in y_predicted]
    y_predicted_new = [min(i,1-epsilon) for i in y_predicted_new]
    y_predicted_new = np.array(y_predicted_new)
    return -np.mean(y_true*np.log(y_predicted_new)+(1-y_true)*np.log(1-y_predicted_new))
```

```python
def gradient_descent(age, affordability, y_true, epochs, loss_thresold):
    w1 = w2 = 1
    bias = 0
    rate = 0.5
    n = len(age)
    for i in range(epochs):
        weighted_sum = w1 * age + w2 * affordability + bias
        y_predicted = sigmoid(weighted_sum)
        loss = log_loss(y_true, y_predicted)

        w1d = (1/n)*np.dot(np.transpose(age),(y_predicted-y_true)) 
        w2d = (1/n)*np.dot(np.transpose(affordability),(y_predicted-y_true)) 

        bias_d = np.mean(y_predicted-y_true)
        w1 = w1 - rate * w1d
        w2 = w2 - rate * w2d
        bias = bias - rate * bias_d

        print (f'Epoch:{i}, w1:{w1}, w2:{w2}, bias:{bias}, loss:{loss}')

        if loss<=loss_thresold:
            break

    return w1, w2, bias
```
